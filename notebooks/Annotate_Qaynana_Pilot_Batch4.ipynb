## Quick label guide

- **Neutral:** no clear emotional coloring.
- **Happy:** cheerful or excited vocal tone.
- **Sad:** subdued, low-energy, or tearful vocal tone.
- **Angry:** tense, irritated, forceful, or sharp vocal tone.
- **Unclear / exclude:** music, noise, overlapping voices, unintelligible speech, or emotion you cannot confidently identify.

Speaker IDs are arbitrary stable tags such as `speaker_01`. Reuse the same tag whenever you recognize the same voice. If unsure, leave it blank; do not guess.
